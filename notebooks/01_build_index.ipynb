# Build BM25 + TF-IDF Index

Build a self-implemented BM25 index and a TF-IDF retriever (both over the same inverted index `term -> {pid: tf}`) from `corpus.jsonl`, then serialize both to `index/`.


```python
import os, sys
for p in [os.path.abspath(".."), os.path.abspath(".")]:
    if os.path.exists(os.path.join(p, "src", "rag")):
        sys.path.insert(0, p)
        break

from src.rag import config
from src.rag.data import loader
from src.rag.retrieval import bm25, tfidf
```

```python
corpus = loader.load_corpus()
print("corpus size:", len(corpus))

index = bm25.build_index(corpus, k1=config.BM25_K1, b=config.BM25_B)
print("num docs:", index.n_docs)
print("num terms (inverted):", len(index.index))
print("avg doc length:", round(index.avgdl, 2))
```

```python
os.makedirs(config.PKG_ROOT / "index", exist_ok=True)
index_path = config.PKG_ROOT / "index" / "bm25.pkl"
bm25.save_index(index, index_path)
print("saved index:", index_path)
```

## Build TF-IDF index

Same inverted index structure, ranked by TF-IDF cosine (sublinear tf + smooth idf + L2 norm). Baseline against BM25.


```python
tfidf_index = tfidf.build_index(corpus)
print("num docs:", tfidf_index.n_docs)
print("num terms (inverted):", len(tfidf_index.index))

tfidf_path = config.PKG_ROOT / "index" / "tfidf.pkl"
tfidf.save_index(tfidf_index, tfidf_path)
print("saved tf-idf index:", tfidf_path)
```
